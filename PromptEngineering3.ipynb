# Exercise 3: Self-Reflection Prompt for Improving Output

## Original Prompt:
Write a short and clear paragraph on why it is important to get an adequate amount of sleep.

## Original Summary:
Getting an adequate amount of sleep is important because it allows the body and mind to recover and function properly. Without enough sleep, people may feel tired, have difficulty focusing, and struggle to perform at their best.

Evaluate the summary using the following criteria:

- Make sure the information provided is accurate.
- Make sure the explanation is clear and easy to understand.
- Make sure the summary is appropriate for a college student.
- Keep the improved summary at most 100 words.
- Keep the improved summary in one clear paragraph.
- Make sure the summary stays focused on why getting adequate sleep is important.

Identify the strengths and weaknesses of the original summary, then rewrite it based on your evaluation.

```python
original_summary = """
Getting an adequate amount of sleep is important because it allows the body
and mind to recover and function properly. Without enough sleep, people may
feel tired, have difficulty focusing, and struggle to perform at their best.
"""

print("ORIGINAL SUMMARY:")
print(original_summary)
```

## Self Reflection:

The summary provided is short and accurate, but I want to talk about specific benefits. I want it to talk about both physicial and mental health while being under 100 words

## Improved Summary:

Getting enough sleep is important for maintaining both physical and mental health. Adequate sleep gives the body time to recover while helping the brain function effectively. It can improve concentration, memory, mood, and energy levels, which can make daily responsibilities easier to manage. Sleep also supports the immune system and helps the body recover from everyday activities. Without sufficient sleep, people may experience fatigue, difficulty concentrating, and reduced performance. Maintaining healthy sleep habits can therefore improve overall well-being and daily functioning.

```python
improved_summary = """
Getting enough sleep is important for maintaining both physical and mental health.
Adequate sleep gives the body time to recover while helping the brain function
effectively. It can improve concentration, memory, mood, and energy levels, which
can make daily responsibilities easier to manage. Sleep also supports the immune
system and helps the body recover from everyday activities. Without sufficient
sleep, people may experience fatigue, difficulty concentrating, and reduced
performance. Maintaining healthy sleep habits can therefore improve overall
well-being and daily functioning.
"""

print("IMPROVED SUMMARY:")
print(improved_summary)
```
