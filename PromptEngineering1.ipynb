Step 1: Classify Issue


Prompt Used:

Act as a customer support assistant for an electronics retailer.

Analyze the following customer message and classify the issue.

Customer message:
"I ordered a black wireless keyboard, but the package I received contains a white wired keyboard. I need the correct item for work next week."

Return:
1. Issue category
2. Urgency level (Low, Medium, or High)
3. A one-sentence description of the problem

Keep the response concise and professional.

```python
customer_message = """
I ordered a black wireless keyboard, but the package I received contains
a white wired keyboard. I need the correct item for work next week.
"""

step1_output = """
Issue Category: Incorrect Item Received
Urgency Level: Medium
Description: The customer received a different keyboard than the one ordered
and needs the correct item before next week.
"""

print("STEP 1 OUTPUT:")
print(step1_output)
```

Step 2: Gather Missing Information

Prompt Used:

Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

[Step 1 output]

Determine what additional information is needed from the customer before the issue can be resolved.

Constraints:

- Ask only for information relevant to verifying the order
- Do not ask for information the customer has already provided
- Do not request sensitive information such as passwords or payment information
- Ask for no more than 2 pieces of information
- Keep the response concise and professional

Return:

Information Needed:

Reason:

```python
# Step 2: Gather Missing Information

step2_prompt = f"""
Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

{step1_output}

Determine what additional information is needed from the customer
before the issue can be resolved.

Constraints:

- Ask only for information relevant to verifying the order
- Do not ask for information the customer has already provided
- Do not request sensitive information such as passwords or payment information
- Ask for no more than 2 pieces of information
- Keep the response concise and professional

Return:

Information Needed:

Reason:
"""

step2_output = """
Information Needed:
1. Order number
2. Photo of the keyboard received

Reason: The order number is needed to verify the original purchase,
and the photo will help confirm that the wrong keyboard was delivered.
"""

print("STEP 2 OUTPUT:")
print(step2_output)
```

Step 3: Propose a Solution

Prompt Used:

Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

[Step 1 output]

The additional information needed was identified as:

[Step 2 output]

Assume the customer has now provided the requested information and the order confirms that they received the wrong keyboard.

Determine the most appropriate solution for the customer.

Constraints:

- Be professional and helpful
- Provide a clear next step for resolving the issue
- Do not guarantee a refund or replacement before the retailer completes its verification process
- Do not ask for information that has already been provided
- Keep the response concise

Return:

Recommended Action:

Customer Response:

```python
# Step 3: Propose a Solution

step3_prompt = f"""
Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

{step1_output}

The additional information needed was identified as:

{step2_output}

Assume the customer has now provided the requested information and
the order confirms that they received the wrong keyboard.

Determine the most appropriate solution for the customer.

Constraints:

- Be professional and helpful
- Provide a clear next step for resolving the issue
- Do not guarantee a refund or replacement before the retailer completes its verification process
- Do not ask for information that has already been provided
- Keep the response concise

Return:

Recommended Action:

Customer Response:
"""

step3_output = """
Recommended Action: Begin the retailer's incorrect-item verification
and return process.

Customer Response: Thank you for providing the requested information.
Since the keyboard you received does not match your order, the next
step is to begin the incorrect-item verification and return process.
Once the item is verified, customer support can provide the available
options for resolving the order.
"""

print("STEP 3 OUTPUT:")
print(step3_output)
```

Step 4: Determine Whether to Escalate

Prompt Used:

Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

[Step 1 output]

The additional information needed was:

[Step 2 output]

The recommended solution was:

[Step 3 output]

Determine whether the issue should be escalated to a customer support supervisor.

Constraints:

- Base the decision on the information from the previous steps
- Escalate if the issue cannot be resolved through the normal return process
- Escalate if the issue requires an exception or additional authorization
- Do not escalate if the normal return process can resolve the issue
- Keep the explanation concise and professional

Return:

Escalation Required: Yes or No

Reason:

```python
# Step 4: Determine Whether to Escalate

step4_prompt = f"""
Act as a customer support assistant for an electronics retailer.

The customer's issue was classified as:

{step1_output}

The additional information needed was:

{step2_output}

The recommended solution was:

{step3_output}

Determine whether the issue should be escalated to a customer
support supervisor.

Constraints:

- Base the decision on the information from the previous steps
- Escalate if the issue cannot be resolved through the normal return process
- Escalate if the issue requires an exception or additional authorization
- Do not escalate if the normal return process can resolve the issue
- Keep the explanation concise and professional

Return:

Escalation Required: Yes or No

Reason:
"""

step4_output = """
Escalation Required: No

Reason: The issue can currently be handled through the retailer's
normal incorrect-item verification and return process. Escalation
would only be necessary if the normal process cannot resolve the issue
or additional authorization is required.
"""

print("STEP 4 OUTPUT:")
print(step4_output)
```
